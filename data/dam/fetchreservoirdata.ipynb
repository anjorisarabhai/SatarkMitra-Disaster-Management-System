```python
import requests
import json
import csv
import os
from datetime import date

def fetch_and_save_reservoir_data():
    """
    Fetches daily reservoir level data from the Data.gov.in API for a specified
    date range and saves it to a CSV file.
    """
    # Define the file name for the API key
    api_key_filename = "api_key.txt"

    # Check if the API key file exists
    if not os.path.exists(api_key_filename):
        print(f"Error: API key file '{api_key_filename}' not found.")
        print("Please create this file and paste your API key inside.")
        return

    # Read the API key from the file
    with open(api_key_filename, 'r') as f:
        api_key = f.read().strip()

    base_url = "https://api.data.gov.in/resource/1fc2148c-fc41-46f5-a364-bdc03f77053f"
    
    start_year = 2018
    end_year = 2025
    
    # Define the directory where the files will be saved.
    # We are now using a hardcoded path to prevent the __file__ error.
    output_dir = r"C:\Users\Dell\Desktop\anjori\SatarkMitra-Disaster-Management-System\data\dam\reservoir_data"
    os.makedirs(output_dir, exist_ok=True)
    
    print("Starting data collection for reservoir levels...")
    
    for year in range(start_year, end_year + 1):
        for month in range(1, 13):
            # Check if the date is in the future and stop if so
            if year > date.today().year or (year == date.today().year and month > date.today().month):
                print("Reached current date. Stopping data collection.")
                break

            params = {
                "api-key": api_key,
                "format": "json",
                "limit": 5000,
                "filters[Year]": str(year),
                "filters[Month]": f"{month:02d}", # Format month with a leading zero
                "filters[Reservoir_name]": "Tehri Reservoir",
                "filters[Agency_name]": "CWC" # Central Water Commission
                # The filters[State] parameter has been removed as it is not supported by this API
            }
            
            try:
                response = requests.get(base_url, params=params)
                response.raise_for_status() # Raises an HTTPError for bad responses
                
                data = response.json()
                records = data.get('records', [])

                if not records:
                    print(f"No records found for Tehri Reservoir, {year}-{month:02d}.")
                    continue
                
                # Use a consistent naming convention for the output files
                output_filename = f"reservoir_level_{year}_{month:02d}.csv"
                output_path = os.path.join(output_dir, output_filename)
                
                # Write the data to a CSV file
                with open(output_path, 'w', newline='', encoding='utf-8') as csvfile:
                    fieldnames = records[0].keys()
                    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
                    writer.writeheader()
                    writer.writerows(records)
                
                print(f"Data for Tehri Reservoir, {year}-{month:02d} successfully saved to {output_path}")

            except requests.exceptions.RequestException as e:
                print(f"An error occurred while fetching data for Tehri Reservoir, {year}-{month:02d}: {e}")
                continue # Continue the loop even if an error occurs

if __name__ == '__main__':
    fetch_and_save_reservoir_data()
```
