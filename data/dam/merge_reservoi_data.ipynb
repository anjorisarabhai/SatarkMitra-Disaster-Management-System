```python
import pandas as pd
import os
import glob

def concatenate_dam_data():
    """
    Finds and concatenates all monthly reservoir level files into a single CSV.
    """
    try:
        # Get the directory where this script is located.
        # This has been updated to a hardcoded path to resolve the '__file__' error.
        script_dir = r"C:\Users\Dell\Desktop\anjori\SatarkMitra-Disaster-Management-System\data\dam"
        data_dir = os.path.join(script_dir, 'reservoir_data')
        
        # Define the pattern to find all monthly data files
        file_pattern = os.path.join(data_dir, 'reservoir_level_*.csv')
        
        # Get a list of all matching file paths
        all_files = glob.glob(file_pattern)
        
        if not all_files:
            print("Error: No 'reservoir_level_*.csv' files found in the 'reservoir_data' directory.")
            return

        print(f"Found {len(all_files)} reservoir data files. Concatenating...")

        # Create a list to hold the DataFrames
        df_list = []
        for file in all_files:
            try:
                # Read each file into a DataFrame
                df = pd.read_csv(file)
                df_list.append(df)
            except Exception as e:
                print(f"Warning: Could not read file {file}. Skipping. Error: {e}")
        
        # Concatenate all DataFrames in the list
        if not df_list:
            print("Error: No valid data could be loaded from the files.")
            return

        combined_df = pd.concat(df_list, ignore_index=True)
        
        print("Data successfully concatenated.")

        # --- Save the combined dataset ---
        output_file_path = os.path.join(script_dir, 'dam_data_combined.csv')
        combined_df.to_csv(output_file_path, index=False)
        
        print(f"Successfully saved combined dam data to: {output_file_path}")

    except Exception as e:
        print(f"An unexpected error occurred: {e}")

if __name__ == '__main__':
    concatenate_dam_data()
```
