```python
import requests
import json
import csv
import os

# Define the file name for the API key
api_key_filename = "api_key.txt"

# Check if the API key file exists
if not os.path.exists(api_key_filename):
    print(f"Error: API key file '{api_key_filename}' not found.")
    print("Please create this file and paste your API key inside.")
    exit()

# Read the API key from the file
with open(api_key_filename, 'r') as f:
    api_key = f.read().strip()

# Base URL for the API endpoint
base_url = "https://api.data.gov.in/resource/6c05cd1b-ed59-40c2-bc31-e314f39c6971"

# Define the parameters for the query, including the higher limit
params = {
    "api-key": api_key,
    "format": "json",
    "limit": 5000,
    "filters[Year]": "2025",
    "filters[State]": "Uttarakhand",
    "filters[Agency_name]": "NRSC VIC MODEL"
}

# Define the full path for the CSV file
csv_save_path = r"C:\Users\Dell\Desktop\anjori\SatarkMitra-Disaster-Management-System\data\rainfall\rainfall_2025.csv"

# Ensure the directory exists
csv_dir = os.path.dirname(csv_save_path)
if not os.path.exists(csv_dir):
    print(f"Creating directory: {csv_dir}")
    os.makedirs(csv_dir)
    print("Directory created successfully.")

try:
    response = requests.get(base_url, params=params)
    response.raise_for_status()

    data = response.json()
    records = data.get('records', [])

    if not records:
        print("No records found in the API response.")
    else:
        with open(csv_save_path, 'w', newline='', encoding='utf-8') as csvfile:
            fieldnames = records[0].keys()
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(records)

        print(f"Data successfully saved to {csv_save_path} with {len(records)} rows")

except requests.exceptions.RequestException as e:
    print(f"An error occurred: {e}")
```
